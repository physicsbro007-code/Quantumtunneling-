# Quantum Tunneling Problem Solver

This notebook reproduces the supplied project using the **thick-barrier approximation**. Default case: electron, 1.0 eV particle energy, 2.0 eV barrier, and 0.5 nm width.

```
import sys
from pathlib import Path
ROOT = Path.cwd()
if (ROOT / 'src').exists():
    sys.path.insert(0, str(ROOT / 'src'))
elif (ROOT.parent / 'src').exists():
    sys.path.insert(0, str(ROOT.parent / 'src'))
from quantum_tunneling import solve_tunneling
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Default calculation

The transmission estimate is \(T \approx e^{-2\kappa a}\), where \(\kappa=\sqrt{2m(V_0-E)}/\hbar\).

```
result = solve_tunneling(particle='Electron', energy_eV=1.0, barrier_eV=2.0, width_nm=0.5)
for key, value in result.items():
    print(f'{key}: {value}')
```

## 2. Transmission versus barrier width

```
widths = np.linspace(0.1, 5.0, 200)
T_width = [solve_tunneling(width_nm=float(w))['transmission_percent'] for w in widths]
plt.figure(figsize=(9,5)); plt.plot(widths, T_width)
plt.xlabel('Barrier width (nm)'); plt.ylabel('Transmission probability (%)')
plt.title('Transmission Probability vs Barrier Width'); plt.grid(True); plt.show()
```

## 3. Transmission versus particle energy

```
energies = np.linspace(0.1, 1.98, 200)
T_energy = [solve_tunneling(energy_eV=float(e))['transmission_percent'] for e in energies]
plt.figure(figsize=(9,5)); plt.plot(energies, T_energy)
plt.xlabel('Particle energy (eV)'); plt.ylabel('Transmission probability (%)')
plt.title('Transmission Probability vs Particle Energy'); plt.grid(True); plt.show()
```

## 4. Potential barrier

```
x = np.linspace(-1.5, 2.0, 500)
V = np.where((x >= 0) & (x <= 0.5), 2.0, 0.0)
plt.figure(figsize=(9,5)); plt.plot(x, V, label='Potential barrier')
plt.axhline(1.0, linestyle='--', label='Particle energy')
plt.xlabel('Position (nm)'); plt.ylabel('Energy (eV)'); plt.title('Potential Barrier')
plt.grid(True); plt.legend(); plt.show()
```

## Model limitation

This is the same simplified exponential thick-barrier model used in the supplied notebook. The reflection estimate is set to \(R\approx1-T\). The plotted potential profile is illustrative; this model does not calculate a fully boundary-matched wavefunction or an exact finite rectangular-barrier transmission coefficient.
